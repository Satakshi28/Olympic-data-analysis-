```python
def most_successful(df, sport):
    temp_df = df.dropna(subset=['Medal'])

    if sport.lower() != 'overall':
        temp_df = temp_df[temp_df['Sport'] == sport]

    # Get top 15 most successful athletes
    top_athletes = temp_df['Name'].value_counts().reset_index().head(15)
    top_athletes.columns = ['Name', 'Medal_Count']

    # Merge with original df to get full athlete details
    merged_df = top_athletes.merge(df, left_on='Name', right_on='Name', how='left')

    return merged_df[['Name', 'Medal_Count', 'ID', 'Sex', 'Age', 'Height', 'Weight', 'Team', 
                      'NOC', 'Games', 'Year', 'Season', 'City', 'Sport', 'Event', 'Medal', 'region']]
```

```python
def most_successful(df, sport):
    temp_df = df.dropna(subset=['Medal'])

    if sport.lower() != 'overall':
        temp_df = temp_df[temp_df['Sport'] == sport]

    vc_df = temp_df['Name'].value_counts().reset_index()
    vc_df.columns = ['Name', 'Medal_Count']
    athlete_info = df[['Name', 'Sport', 'region']].drop_duplicates(subset=['Name'])
    merged_df = vc_df.head(15).merge(athlete_info, on='Name', how='left')

    return merged_df
```
